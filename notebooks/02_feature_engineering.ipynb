Load Cleaned Dataset

```python
import pandas as pd 
import numpy as np 
```

```python
train = pd.read_csv("../data/processed/train_cleaned.csv")

print("Dataset loaded")
print(train.shape)
```

Separate Target Variable

```python
X = train.drop(columns='isFraud')
y =train['isFraud']

print(X.shape)
print(y.shape)
```

Check Categorical Columns

```python
cat_cols = train.select_dtypes(include='object').columns
print("Categorical Columns:",len(cat_cols))
```

Encode Categorical Features

```python
from sklearn.preprocessing import LabelEncoder

le = LabelEncoder()

for col in cat_cols:
    X[col] = le.fit_transform(X[col])

print("Categorical columns encoded successfully")
```

Combine Back

```python
train_encoded = X.copy()
train_encoded["isFraud"] = y

print(train_encoded.shape)
```

Save Feature Engineered Dataset

```python
train_encoded.to_csv("../data/processed/train_featured.csv", index=False)

print("Featured dataset saved successfully")
```
